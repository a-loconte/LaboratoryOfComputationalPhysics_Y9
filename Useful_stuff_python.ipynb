```python
### Esempi sulle copie
import copy
a = [[0, 1], 3,[5, 6]]

alias = a
shallow = list(a)
shallowcp = copy.copy(a)
deep = copy.deepcopy(a)

alias[0].append('alias')  ##this changes a
shallow[0].append('shallow')  ##this changes a
shallowcp[0].append('shallowcopy')  ##this changes a
deep[0].append('deep')  ##this does NOT change a

print(a)
print(shallow)
print(shallowcp)
print(deep)
```
