3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
### TRY 1
def dist(a, b):
    dx = ((b[0] - a[0])**2)
    dy = ((b[1] - a[1])**2)
    de = (dx + dy)**(1/2)
    return de

x = (3,1)
y = (2,4)
dist(x, y)
```

```python
## Metodo 2
def diste(x,y):
    return sum((a-b)**2 for a,b in zip(x,y)) ** 0.5
    #zip(a,b) prende a[0],b[0] e poi a[1],b[1]

diste((3,0),(0,4))
    
##prova mia
a = 1,2,3
b = 1,2
c = zip(a,b)
print (tuple(c)) ##zip arriva fino a dove IL PIù CORTO permette!!

##altra prova
d = ((x) for x in zip(a,b))
print(tuple(d))  ##se la tupla ha due var e ne metto una sola, quella solo si riferirà alla tupla intera, non all'elemento interno
                 ##se metto numero di var uguali, allora andrà elemento per elemento e ho più libertà (risultato uguale)
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
### TRY 1
def square(x):
    return x**2

def cube(x):
    return x**3

def sixth(x):
    a = square(x)
    b = cube(x)
    return (a*b*x)

sixth(2)

```

```python
### modo figo
def square(x):
    return x**2

def cube(x):
    return x**3

def sixth(x):
    return square(cube(x))

sixth(2)

###puoi sempre fare la funzione di una funzione che ha un return!!!
```

7\. Cubes

Create a list of the cubes of x for x in *[0, 10]* using:

a) a for loop

b) a list comprehension

```python
### TRY 1
def f_cubes():
    a = []
    for i in range(0, 11):
        a.append(i**3)
    return a


def lc_cubes():
    cubes_a = [i**3 for i in range(0,11)]
    return cubes_a

print(f_cubes(), lc_cubes())

```

```python
### provo con la funzione map
def lc_cubes():
    cubes_a = list(map(lambda i: i**3, range(0, 11)))
    return cubes_a
lc_cubes()

### figo, ma occhio che se lo fai così NON PUOI fare una list comprehension
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
### TRY 1
def find_p():
    z = []
    for a in range(1, 101):
        for b in range(1, 101):
            for c in range(1, 101):
                if a**2+b**2 == c**2:
                    z.append((a, b, c))
    return z

find_p()

### vari problemi
# 1) richiede una tupla, io ho fatto una lista di tuple
# 2) devo fare una NESTED LIST COMPREHENSION
# 3) le permutazioni sono ancora presenti (ho sia (3, 4, 5) che (4, 3, 5))

```

```python
### TRY 2 per risolvere questi problemi
def find_p():
    z = tuple((a, b, c)
             for a in range(1, 100)
             for b in range(1, 100)
             for c in range(1, 100)
             if a**2+b**2 == c**2 and a>b and b>c)   ## non lo puoi scrivere così, si blocca alla prima iterazione
    return z

find_p()
```

```python
### TRY 3
def find_p():
    z = tuple((a, b, c)
             for a in range(1, 100)
             for b in range(a, 100)
             for c in range(b, 100)
             if a**2+b**2 == c**2)   ## non lo puoi scrivere così, si blocca alla prima iterazione
    return z

find_p()


### IDEA: controlla sempre se
1) puoi avere relazioni fra quantità che hai già inserito
2) usare funzioni già scritte ma inserendo variabili ==> simmetrie (posso chiamarle così? forse no)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
### TRY 1 (somma entrate normalizzata ad 1)
def norm(a, b, c):
    s = 0
    for j in (a, b, c):
        s = s + j
    tn = (a/s, b/s, c/s)
    #for i in (a, b, c):
        #tn = tn + tuple(i/s)
    return tn

norm(1, 2, 3)

### problemi:
#
```

```python
### metodo elegante
def norm(t):
    s = 0
    for j in t:
        s = s + j
        
        
    return tuple(i/s for i in t)

t = (3, 5, 6)
norm(t)
```
