````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
### TRY 1 with recursion
def fact(x):
    if x == 0:
        return 1
    else:
        return x*fact(x-1)

fact(5)
````

````python
### TRY 2
def fact_norec(x):
    a = 1
    for i in range(1, x):
        a = a*x   ## non puoi usare x, i a cosa ti serve poi (a fare solo la stessa operazione troppe volte)
    return a

fact_norec(5)
````

````python
### TRY 2
def fact2(x):
    a = 1
    for i in range(1, x+1):
        a = a * i
    return a

fact2(5)

````
